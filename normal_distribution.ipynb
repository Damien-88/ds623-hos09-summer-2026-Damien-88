```python
# Import Libraries
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb

# Create the distribution of Seattle latte prices
seattle_data = np.arange(0, 10, 0.01)
seattle_pdf = norm.pdf(seattle_data, loc = 4.7, scale = 1)

# Visualize the distributionsb.set_style("whitegrid")
sb.lineplot(x = seattle_data, y = seattle_pdf, color = "green")
plt.xlabel("Seattle Latte Prices")
plt.ylabel("Probability Density")
```

```python
# Create the distribution of Bellevue latte prices
bellevue_data = np.arange(0, 10, 0.01)
bellevue_pdf = norm.pdf(bellevue_data, loc = 5.2, scale = 0.7)

# Visualize the distribution
sb.set_style("whitegrid")
sb.lineplot(x = bellevue_data, y = bellevue_pdf, color = "blue")
plt.xlabel("Bellevue Latte Prices")
plt.ylabel("Probability Density")
```

```python
# Calculate the probability of a Seattle latte costing less than $4.50
seattle_prob_450 = round(norm(loc = 4.7, scale = 1).cdf(4.5), 4)
print("Probability of a random Seattle coffee shop's latte costing less than $4.50:")
print(seattle_prob_450, "or", "{:.2f}".format(seattle_prob_450 * 100) + "%\n")

# Set the x range for a cost up to $4.50 and calculate the pdf
seattle_px = np.arange(0, 4.5, 0.01)
seattle_px_pdf = norm.pdf(seattle_px, loc = 4.7, scale = 1)

# Plot the probability function including the area under the curve
sb.set_style("whitegrid")
sb.lineplot(x = seattle_data, y = seattle_pdf, color = "green")
plt.fill_between(seattle_px, seattle_px_pdf, color = "red", alpha = 0.5)
plt.xlabel("Seattle Latte Prices < $4.50")
plt.ylabel("Probability Density")
```

```python
# Calculate the probability of a Seattle latte costing between $4 and $6
seattle_prob_4 = norm(loc = 4.7, scale = 1).cdf(4)
seattle_prob_6 = norm(loc = 4.7, scale = 1).cdf(6)

seattle_prob_4to6 = round(seattle_prob_6 - seattle_prob_4, 4)

print("Probability of a random Seattle coffee shop's latte costing between $4 and $6:")
print(seattle_prob_4to6, "or", "{:.2f}".format(seattle_prob_4to6 * 100) + "%\n")

# Set the x range for a cost between $4.00 and $6.00 and calculate the pdf
seattle_px = np.arange(4, 6, 0.01)
seattle_px_pdf = norm.pdf(seattle_px, loc = 4.7, scale = 1)

# Plot the probability function including the area under the curve
sb.set_style("whitegrid")
sb.lineplot(x = seattle_data, y = seattle_pdf, color = "green")
plt.fill_between(seattle_px, seattle_px_pdf, color = "red", alpha = 0.5)
plt.xlabel("Seattle Latte Prices \$4 and \$6")
plt.ylabel("Probability Density")
```
